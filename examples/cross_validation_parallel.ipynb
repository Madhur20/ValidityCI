# Parallel cross-validation support-boundary benchmark
Uses worker processes so preprocessing lineage must be reported as unsupported rather than passed.

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GroupKFold, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

data = pd.read_csv('customer_churn.csv')
X = data[['monthly_spend', 'support_tickets']]
y = data['churned']
entities = data['customer_id']
pipeline = Pipeline([
    ('scale', StandardScaler()),
    ('model', LogisticRegression(random_state=42)),
])
scores = cross_val_score(
    pipeline, X, y, groups=entities, cv=GroupKFold(n_splits=2),
    scoring='accuracy', n_jobs=2
)
accuracy = float(scores.mean())
accuracy
```
