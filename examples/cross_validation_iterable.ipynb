# Iterable cross-validation lineage benchmark
Exercises explicit train/test index pairs without serializing those indices.

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

data = pd.read_csv('customer_churn.csv')
X = data[['monthly_spend', 'support_tickets']]
y = data['churned']
entities = data['customer_id']
even = np.arange(0, len(data), 2)
odd = np.arange(1, len(data), 2)
folds = [(even, odd), (odd, even)]
pipeline = Pipeline([
    ('scale', StandardScaler()),
    ('model', LogisticRegression(random_state=42)),
])
scores = cross_val_score(
    pipeline, X, y, groups=entities, cv=folds, scoring='accuracy'
)
accuracy = float(scores.mean())
accuracy
```
