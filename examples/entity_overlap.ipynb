# Entity-overlap benchmark
Intentionally performs a row-level split over repeated customer entities.

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split

data = pd.read_csv('entity_overlap.csv')
X = data[['monthly_spend', 'support_tickets']]
y = data['churned']
entities = data['customer_id']
X_train, X_test, y_train, y_test, entity_train, entity_test = train_test_split(
    X, y, entities, test_size=0.5, random_state=42, stratify=y
)
model = LogisticRegression(random_state=42)
model.fit(X_train, y_train)
predictions = model.predict(X_test)
accuracy = accuracy_score(y_test, predictions)
accuracy
```
